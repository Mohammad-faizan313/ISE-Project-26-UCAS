# ISE-A1 Evidence Summary\nThis notebook reads the versioned evidence ledger and produces deterministic checks used to support the report. It does not scrape the web or use patient data.

```python
from pathlib import Path\nimport csv\nfrom collections import Counter\n\nROOT = Path('..').resolve()\nCSV_PATH = ROOT / 'evidence' / 'evidence_ledger.csv'\nwith CSV_PATH.open('r', encoding='utf-8', newline='') as f:\n    rows = list(csv.DictReader(f))\nprint(f'Loaded {len(rows)} evidence records from {CSV_PATH}')
```

```python
source_types = Counter(r['SourceType'] for r in rows)\nyears = sorted({r['Year'] for r in rows})\nprint('Source types:', dict(source_types))\nprint('Publication years:', years)\nprint('Quantitative reference records:', [r['ID'] for r in rows if r['MetricOrFinding'] and any(ch.isdigit() for ch in r['MetricOrFinding'])])
```
